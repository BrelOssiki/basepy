Exercice 5
Écrivez une fonction en Python qui prend en entrée un texte en anglais ou en français et qui
calcule la fréquence d'apparition de chaque lettre. La fonction devrait retourner un
dictionnaire où les clés sont les lettres de l'alphabet et les valeurs sont les fréquences
correspondantes.

```python
def frequence_lettres(texte):
    # Convertissons le texte en minuscules pour faciliter le traitement
    texte = texte.lower()

    # un dictionnaire pour stocker les fréquences de chaque lettre
    alphabet="abcdefghijklmnopqrstuvwxyz"
    frequences = {lettre: 0 for lettre in alphabet }

    # Parcourons chaque lettre du texte et mettre à jour les fréquences
    for lettre in texte:
        if lettre in frequences:
            frequences[lettre] += 1

    # Calculs des fréquences relatives
    nombre_total_lettres = sum(frequences.values())
    for lettre in frequences:
        frequences[lettre] /= nombre_total_lettres

    return frequences

#Voici un exemple d'utilisation de cette fonction :
texte1 = "Bonjour tout le monde, vous allez tous bien les gars!"
texte2 = "Hello, World! Money will change me, i told you so there won't be no confusion"
print(frequence_lettres(texte1))
print(frequence_lettres(texte2))

```
