1. Écrire une fonction qui prend deux listes en entrée et renvoie une nouvelle liste contenant les éléments communs aux deux listes :

```python
def elements_communs(liste1, liste2):
    commun=[element for element in liste1 if element in liste2]
    return commun

# Exemple d'utilisation :
liste1 = [1, 2, 3, 4, 5]
liste2 = [3, 4, 5, 6, 7]

print(elements_communs(liste1, liste2))  
```

2. Créer une fonction qui prend un nombre en entrée et renvoie "pair" ou "impair" :

```python
def pair_ou_impair(nombre):
    if nombre % 2 != 0:
        return "impair"
    else:
        return "pair"

# Exemple d'utilisation :
print(pair_ou_impair(4))  # Renvoie "impair"
```

3. Écrire une fonction qui prend une liste de nombres en entrée et renvoie la somme de tous les éléments :

```python
def somme_elements(liste):
    somme=sum(liste)
    return somme

# Exemple d'utilisation :
print(somme_elements([1, 2, 3, 4, 5, 6])) 
```

4. Créer une fonction qui prend une liste de mots en entrée et renvoie une nouvelle liste contenant la première lettre de chaque mot :

```python
def premieres_lettres(liste_mots):
    lettre=[mot[0] for mot in liste_mots]
    return lettre

# Exemple d'utilisation :
mots = ["chat", "chien", "souris", "éléphant"]
print(premieres_lettres(mots)) 
```
