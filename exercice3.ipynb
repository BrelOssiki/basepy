
1. Créer une liste de nombres et écrire une fonction qui renvoie la médiane de cette liste :

```python
def mediane(liste):
    liste_triee = sorted(liste)
    longueur = len(liste_triee)
    if longueur % 2 == 0:
        longueur_paire= (liste_triee[longueur // 2 - 1] + liste_triee[longueur // 2]) / 2
        return longueur_paire
    else:
        longeur_impair= liste_triee[longueur // 2]
        return longeur_impair

# Exemple d'utilisation :
nombres = [4, 7, 1, 9, 3, 5]
print(mediane(nombres))  
```

2. Écrire une fonction qui prend une liste de chaînes de caractères en entrée et renvoie une nouvelle liste contenant seulement les chaînes de plus de 5 caractères :

```python
def plus_de_cinq_caracteres(liste):
    return [mot for mot in liste if len(mot) > 5]

# Exemple d'utilisation :
mots = ["Photographie", "Synthetisation", "Exposition", "echo", "Beta"]
print(plus_de_cinq_caracteres(mots))  
```

3. Créer une liste de nombres et écrire un programme qui calcule la moyenne de ces nombres sans utiliser la fonction sum :

```python
def moyenne(liste):
    total = 0
    for nombre in liste:
        total += nombre
    return total / len(liste)

# Exemple d'utilisation :
nombres = [10, 20, 30, 40, 50]
print(moyenne(nombres))  # Renvoie 30.0
```

4. Écrire une fonction qui prend une liste en entrée et renvoie la liste inversée :

```python
def liste_inverse(liste):
    return liste[::-1]

# Exemple d'utilisation :
elements = [1, 2, 3, 4, 5]
print(liste_inverse(elements))  # Renvoie [5, 4, 3, 2, 1]
```

5. Créer une liste de noms et afficher les noms qui contiennent la lettre "a" :

```python
def noms_avec_a(liste):
    return [nom for nom in liste if 'a' in nom]

# Exemple d'utilisation :
noms = ["Alice", "Bob", "Charlie", "David", "Eve", "Olivia"]
print(noms_avec_a(noms))  # Renvoie ['Alice', 'David', 'Olivia']
```
