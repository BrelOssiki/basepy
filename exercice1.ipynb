1. Créer une liste de nombres de 1 à 10 et afficher chaque nombre multiplié par 2 :

```python
nombres = list(range(1, 11))
for nbr in nombres:
    print(nbr * 2)
```

Créer une liste de noms et afficher la longueur de chaque nom en utilisant une boucle :

```python
noms = ["Chacko", "Romaric", "Neylove", "Africain"]
for prenom in noms:
    print(len(prenom))
```

3. Écrire une fonction qui prend une liste en entrée et renvoie une nouvelle liste contenant le carré de chaque élément 

```python
def carre_liste(liste):
    return [x ** 2 for x in liste]

ma_liste = [1, 2, 3, 4, 5, 6, 7]
nbr_carre = carre_liste(ma_liste)
print(nbr_carre)
```

Créer une liste de nombres aléatoires et afficher seulement les nombres qui sont supérieurs à 5 

```python
import random

random_nombres = [random.randint(1, 10) for _ in range(10)]
superieure=[nb for nb in random_nombres if nb > 5]
print(superieure)
```

5. Écrire un programme qui prend une liste de chaînes de caractères en entrée et affiche la plus longue 

```python
liste = ["pomme", "banane", "kiwi", "fraise", "orange", "pastèque"]
plus_longue = max(liste, key=len)
print(plus_longue)
```
