1. Créer une fonction qui prend une liste en entrée et renvoie la liste sans les éléments en doublon :

```python
def doublons(liste):
    return list(set(liste))

# d'utilisation :
elements = [1, 2, 3, 2, 4, 5, 3, 6]
print(doublons(elements))  
```

2. Créer une fonction qui génère une liste de nombres premiers jusqu'à un nombre donné 

```python
def nombres_premiers(maximum):
    nombres = []
    for nombre in range(2, maximum + 1):
        premier = True
        for diviseur in range(2, int(nombre ** 1/2) + 1):
            if nombre % diviseur == 0:
                premier = False
                break
        if premier:
            nombres.append(nombre)
    return nombres

# utilisation :
print(nombres_premiers(20))  
```


3. Écrire un programme qui utilise la fonction map pour multiplier chaque élément d'une liste par 3 :

```python
def multiplier(nombre):
    return nombre * 3

# Exemple d'utilisation avec map :
nombres = [1, 2, 3, 4, 5]
resultat = list(map(multiplier, nombres))
print(resultat)  
```
